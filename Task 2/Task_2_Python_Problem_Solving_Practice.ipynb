## Q8. Find the Most Frequent Character

### 1. Observe
Input: `banana`; expected output: `a`.

### 2. Understand
Find the character with the highest frequency, ignoring spaces. Ties go to the first encountered character.

### 3. Break
Build character counts in insertion order, then select the first character with the greatest count.

### 4. Logic
Count each non-space character. Scan the counts and update the answer only for a strictly higher frequency.

### 5. Code
Run the code cell below in Jupyter Notebook.

### 6. Test
The code tests the task's example and an additional case. Expected outputs are shown in the comments after the code.

```python
text = "banana"
counts = {}

for character in text:
    if character != " ":
        counts[character] = counts.get(character, 0) + 1

most_frequent = None
highest_count = 0
for character, count in counts.items():
    if count > highest_count:
        most_frequent = character
        highest_count = count

print(most_frequent)

# Additional test case: a and b tie; a appears first
text = "ab ba"
counts = {}
for character in text:
    if character != " ":
        counts[character] = counts.get(character, 0) + 1
most_frequent = None
highest_count = 0
for character, count in counts.items():
    if count > highest_count:
        most_frequent = character
        highest_count = count
print(most_frequent)

# Expected output:
# a
# a
```

## Q9. Find the Least Frequent Character

### 1. Observe

### 2. Understand
Find the character with the lowest frequency, ignoring spaces. Ties go to the first encountered character.

### 3. Break

### 5. Code
Run the code cell below in Jupyter Notebook.

## Q16. Find the Most Frequent Word in a Sentence

### 1. Observe
Input: `Python is easy and Python is powerful`; expected output: `python`.

Find the most common word, treating uppercase and lowercase as equal. A tie goes to the word first encountered.

## Q20. Find the Longest Word That Appears Only Once

### 1. Observe
Input: `Python makes coding useful Python`; expected output: `coding`.

### 2. Understand
Find the longest word appearing exactly once, treating uppercase and lowercase as equal. A tie goes to the first qualifying word.

### 3. Break
Normalize words, count each word, then compare lengths only among words with frequency one.

### 4. Logic
Scan words in order. If a word appears once and is longer than the current answer, update it. Strict comparison preserves the first tie.

### 5. Code
Run the code cell below in Jupyter Notebook.

### 6. Test
The code tests the task's example and an additional case. Expected outputs are shown in the comments after the code.

```python
sentence = "Python makes coding useful Python"
words = sentence.lower().split()
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

longest_unique = None
for word in words:
    if counts[word] == 1:
        if longest_unique is None or len(word) > len(longest_unique):
            longest_unique = word

print(longest_unique)

# Additional test case: "apple" and "grape" tie; apple appears first
sentence = "apple grape pear"
words = sentence.lower().split()
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
longest_unique = None
for word in words:
    if counts[word] == 1:
        if longest_unique is None or len(word) > len(longest_unique):
            longest_unique = word
print(longest_unique)

# Expected output:
# coding
# apple
```
