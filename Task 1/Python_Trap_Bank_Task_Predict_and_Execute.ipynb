#Q35

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

- A shallow copy duplicates only the outer list.
- The inner lists are shared, so changing an inner element through b also changes the nested list visible through a.

#Q36

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```

- deepcopy() recursively copies the nested lists, so changing b's inner element does not modify a

#Q37

```python
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
```

- list.copy() makes a shallow copy. The outer lists are different, but both contain a reference to the same inner list [20].

#Q38

```python
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
```

- deepcopy() creates independent nested lists, so appending to a's inner list does not affect b's copy.

#Q39

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
```

- The shallow copy creates a new outer list, but its inner lists are references to the same objects as those in a.
